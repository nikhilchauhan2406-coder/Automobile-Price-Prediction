```python
import pandas as pd
import os
data_path=os.path.join(os.getcwd(),"Data")
file_path=os.path.join(data_path,"auto_imports.csv")
column_names=["symboling","normalized-losses","make","fuel-type","aspiration","num-of-doors","body-style","drive-wheels","engine-location","wheel-base","length","width","height","curb-weight","engine-type","num-of-cylinders","engine-size","fuel-system","bore","stroke","compression-ratio","horsepower","peak-rpm","city-mpg","highway-mpg","price",]
df=pd.read_csv(file_path,header=None,names=column_names,na_values="?")
print("Dataset loaded with correct colum names!")
print("Dataset shape:",df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst five rows:")
print(df.head())
print("\nPrice column exists:","price" in df.columns)
```

```python
#Step 20:Build the preprocessing pipeline
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
#Separate input features & target variable
X=df.drop(columns=["price"])
Y=pd.to_numeric(df["price"],errors="coerce")
#Remove rows where the target price is missing or invalid
valid_rows=Y.notna()
X=X.loc[valid_rows].copy()
Y=Y.loc[valid_rows].copy()
#Identify numerical & categorical columns
numerical_features=X.select_dtypes(include=["number"]).columns.tolist()
categorical_features=X.select_dtypes(include=["object","category","bool"]).columns.tolist()
print("Numerical features")
print(numerical_features)
print("\nCategorical features:")
print(categorical_features)
#Split data before fitting preprocessing transformations
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.20,random_state=42)
#Numerical preprocessing
numerical_pipeline=Pipeline(steps=[("imputer",SimpleImputer(strategy="median")),("scaler", StandardScaler())])
#Categorical preprocessing
categorical_pipeline=Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore"))])
#Combine both preprocessing pipelines
preprocessor=ColumnTransformer(transformers=[("num",numerical_pipeline,numerical_features),("cat",categorical_pipeline,categorical_features)])
print("\nPreprocessing pipeline created successfully.")
print("Training rows:",len(X_train))
print("Testing rows:",len(X_test))
print("CHECKING PREPROCESSOR:")
print("Preprocessor available:","preprocessor"in globals())
```

```python
print("Checking dataframe...")
print("df exists:","df" in globals())
print("data exists:","data" in globals())
```

```python
#Step 21:Apply preprocessing to training and testing data
X_train_processed=preprocessor.fit_transform(X_train)
X_test_processed=preprocessor.transform(X_test)
print("Preprocessing completed successfully!")
print("Training data shape:",X_train_processed.shape)
print("Testing data shape:",X_test_processed.shape)
```

```python
print ("Checking preprocessing variables...")
print ("X_train_processed exists:",'X_train_processed' in globals())
print ("X_test_processed exists:",'X_test_processed' in globals())
print ("Preprocessor exists:",'preprocessor' in globals())
```

```python
#Step 22:Train Linear Regression Model
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score 
import numpy as np
linear_model=LinearRegression()
linear_model.fit(X_train_processed,Y_train)
linear_predictions=linear_model.predict(X_test_processed)
print("Linear Regression Model Trained Successfully!")
print("Mean Absolute Error(MAE):",mean_absolute_error(Y_test,linear_predictions))
print("Root Mean Squared Error(RMSE):",np.sqrt(mean_squared_error(Y_test,linear_predictions)))
print("R2 Score:",r2_score(Y_test,linear_predictions))
```

```python
#Step 23:Decision Tree Regression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
dt_model=DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train_processed,Y_train)
dt_predictions=dt_model.predict(X_test_processed)
print("Decision Tree Model Trained Successfully!")
print("MAE:",mean_absolute_error(Y_test,dt_predictions))
print("RMSE:",np.sqrt(mean_squared_error(Y_test,dt_predictions)))
print("R2 Score:",r2_score(Y_test,dt_predictions))
```

```python
#Step 24:Random Forest Regression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
rf_model=RandomForestRegressor(n_estimators=200,random_state=42)
rf_model.fit(X_train_processed,Y_train)
rf_predictions=rf_model.predict(X_test_processed)
print("Random Forest Model Trained Successfully!")
print("MAE:",mean_absolute_error(Y_test,rf_predictions))
print("RMSE:",np.sqrt(mean_squared_error(Y_test,rf_predictions)))
print("R2 Score:",r2_score(Y_test,rf_predictions))
```

```python
#Step 25:Gradient Boosting Regression
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
gb_model=GradientBoostingRegressor(random_state=42)
gb_model.fit(X_train_processed,Y_train)
gb_predictions=gb_model.predict(X_test_processed)
print("Gradient Boosting Model Trained Successfully!")
print("MAE:",mean_absolute_error(Y_test,gb_predictions))
print("RMSE:",np.sqrt(mean_squared_error(Y_test,gb_predictions)))
print("R2 Score:",r2_score(Y_test,gb_predictions))
```

```python
#Step 26:Comapre model preference
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

results=pd.DataFrame({"Model":["Linear Regression","Decision Tree","Random Forest","Gradient Boosting"],
"MAE":[mean_absolute_error(Y_test,linear_predictions),mean_absolute_error(Y_test,dt_predictions),mean_absolute_error(Y_test,rf_predictions),mean_absolute_error(Y_test,gb_predictions)],
"RMSE":[np.sqrt(mean_squared_error(Y_test,linear_predictions)),np.sqrt(mean_squared_error(Y_test,dt_predictions)),
np.sqrt(mean_squared_error(Y_test,rf_predictions)),np.sqrt(mean_squared_error(Y_test,gb_predictions))],
"R2 Score":[r2_score(Y_test,linear_predictions),r2_score(Y_test,dt_predictions),r2_score(Y_test,rf_predictions),r2_score(Y_test,gb_predictions)]})
print(results.sort_values(by="MAE"))
```

```python
#Step 27:Identify the best model by MAE
best_model_name=results.loc[results["MAE"].idxmin(),"Model"]
print("Best Model Based on MAE:",best_model_name)
```

```python
#Step 28:Random Forest Hyperparameter Tuning
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
param_grid={"n_estimators":[100,200],"max_depth":[None,5,10],"min_samples_split":[2,5]}
grid_search=GridSearchCV(estimator=RandomForestRegressor(random_state=42),param_grid=param_grid,
scoring="neg_mean_absolute_error",cv=5,n_jobs=-1)
grid_search.fit(X_train_processed,Y_train)
print("Best Parameters:",grid_search.best_params_)
print("Best Cross-Validation MAE:",-grid_search.best_score_)
```
